```python
import math

def print_board(board):
    print("\n")
    for row in [board[i*3:(i+1)*3] for i in range(3)]:
        print(' | '.join(row))
        print("-" * 9)

def check_winner(board, player):
    win_conditions = [
        (0,1,2), (3,4,5), (6,7,8), # Rows
        (0,3,6), (1,4,7), (2,5,8), # Cols
        (0,4,8), (2,4,6)           # Diagonals
    ]
    return any(board[i] == board[j] == board[k] == player for i, j, k in win_conditions)

def is_draw(board):
    return ' ' not in board

def minimax(board, depth, is_maximizing):
    if check_winner(board, 'O'): return 10 - depth # AI wins
    if check_winner(board, 'X'): return depth - 10 # Human wins
    if is_draw(board): return 0

    if is_maximizing:
        best_score = -math.inf
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'O'
                score = minimax(board, depth + 1, False)
                board[i] = ' '
                best_score = max(score, best_score)
        return best_score
    else:
        best_score = math.inf
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'X'
                score = minimax(board, depth + 1, True)
                board[i] = ' '
                best_score = min(score, best_score)
        return best_score

def get_best_move(board):
    best_score = -math.inf
    move = -1
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            score = minimax(board, 0, False)
            board[i] = ' '
            if score > best_score:
                best_score = score
                move = i
    return move

def play_game():
    board = [' ' for _ in range(9)]
    print("Tic-Tac-Toe Minimax! You are 'X', AI is 'O'.")
    print("Positions are 0-8 (left to right, top to bottom).")

    while True:
        print_board(board)

        # Human Move
        try:
            human_move = int(input("Enter your move (0-8): "))
            if board[human_move] != ' ':
                print("Invalid move! Try again.")
                continue
            board[human_move] = 'X'
        except (ValueError, IndexError):
            print("Please enter a valid number between 0 and 8.")
            continue

        if check_winner(board, 'X'):
            print_board(board)
            print("You win! (Wait, that's impossible...)")
            break
        elif is_draw(board):
            print_board(board)
            print("It's a draw!")
            break

        # AI Move
        print("AI is thinking...")
        ai_move = get_best_move(board)
        board[ai_move] = 'O'

        if check_winner(board, 'O'):
            print_board(board)
            print("AI wins!")
            break
        elif is_draw(board):
            print_board(board)
            print("It's a draw!")
            break

if __name__ == "__main__":
    play_game()
```

```python
import os
import time
from collections import deque
import heapq

# Configuration
DELAY = 0.05 # Animation speed
GRID = [
    list("S  #      "),
    list(" # # #### "),
    list(" #      # "),
    list(" #### # # "),
    list("      #  E")
]

ROWS, COLS = len(GRID), len(GRID[0])

def find_start_end(grid):
    start = end = None
    for r in range(ROWS):
        for c in range(COLS):
            if grid[r][c] == 'S': start = (r, c)
            if grid[r][c] == 'E': end = (r, c)
    return start, end

def print_grid(grid, title):
    os.system('cls' if os.name == 'nt' else 'clear')
    print(f"--- {title} ---")
    for row in grid:
        print(" ".join(row))
    time.sleep(DELAY)

def get_neighbors(r, c):
    neighbors = []
    for dr, dc in [(-1,0), (1,0), (0,-1), (0,1)]: # Up, Down, Left, Right
        nr, nc = r + dr, c + dc
        if 0 <= nr < ROWS and 0 <= nc < COLS and GRID[nr][nc] != '#':
            neighbors.append((nr, nc))
    return neighbors

def reconstruct_path(came_from, current, grid):
    while current in came_from and came_from[current] is not None:
        r, c = current
        if grid[r][c] not in ('S', 'E'):
            grid[r][c] = '*'
        current = came_from[current]

def bfs():
    grid = [row[:] for row in GRID]
    start, end = find_start_end(grid)
    queue = deque([start])
    came_from = {start: None}

    while queue:
        curr = queue.popleft()
        if curr == end:
            reconstruct_path(came_from, curr, grid)
            print_grid(grid, "BFS Path Found")
            return

        r, c = curr
        if grid[r][c] not in ('S', 'E'): grid[r][c] = '.'
        print_grid(grid, "BFS Searching...")

        for n in get_neighbors(r, c):
            if n not in came_from:
                queue.append(n)
                came_from[n] = curr

def dfs():
    grid = [row[:] for row in GRID]
    start, end = find_start_end(grid)
    stack = [start]
    came_from = {start: None}

    while stack:
        curr = stack.pop()
        if curr == end:
            reconstruct_path(came_from, curr, grid)
            print_grid(grid, "DFS Path Found")
            return

        r, c = curr
        if grid[r][c] not in ('S', 'E'): grid[r][c] = '.'
        print_grid(grid, "DFS Searching...")

        for n in get_neighbors(r, c):
            if n not in came_from:
                stack.append(n)
                came_from[n] = curr

def heuristic(a, b):
    # Manhattan distance
    return abs(a[0] - b[0]) + abs(a[1] - b[1])

def a_star():
    grid = [row[:] for row in GRID]
    start, end = find_start_end(grid)

    open_set = []
    heapq.heappush(open_set, (0, start))
    came_from = {start: None}
    g_score = {start: 0}

    while open_set:
        _, curr = heapq.heappop(open_set)

        if curr == end:
            reconstruct_path(came_from, curr, grid)
            print_grid(grid, "A* Path Found")
            return

        r, c = curr
        if grid[r][c] not in ('S', 'E'): grid[r][c] = '.'
        print_grid(grid, "A* Searching...")

        for n in get_neighbors(r, c):
            tentative_g = g_score[curr] + 1
            if n not in g_score or tentative_g < g_score[n]:
                came_from[n] = curr
                g_score[n] = tentative_g
                f_score = tentative_g + heuristic(n, end)
                heapq.heappush(open_set, (f_score, n))

if __name__ == "__main__":
    print("Starting visualizer in 2 seconds...")
    time.sleep(2)

    bfs()
    time.sleep(2)
    dfs()
    time.sleep(2)
    a_star()
```
